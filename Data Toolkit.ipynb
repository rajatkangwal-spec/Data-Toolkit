# Some Extra Examples

## Example 1: NumPy Array

NumPy can be used to store numbers in an array and do calculations easily.

```python
import numpy as np

numbers = np.array([10, 20, 30, 40])
print(numbers)
print("Sum =", numbers.sum())
print("Mean =", numbers.mean())
```

## Example 2: Pandas DataFrame

A DataFrame looks like a small table. We can select columns and calculate values from it.

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Ravi", "Aman", "Neha"],
    "Marks": [65, 78, 88]
})

print(df)
print("Average marks =", df["Marks"].mean())
```

## Example 3: Simple Seaborn Plot

Seaborn can be used to make simple statistical graphs from data.

```python
import seaborn as sns
import matplotlib.pyplot as plt

marks = [45, 55, 60, 70, 72, 80, 85, 90]

sns.histplot(marks)
plt.title("Simple Marks Distribution")
plt.show()
```

## 3. Create a scatter plot using Matplotlib.

```python
import matplotlib.pyplot as plt
x=[1,2,3,4,5]
y=[2,4,3,7,6]
plt.scatter(x,y)
plt.xlabel("X values")
plt.ylabel("Y values")
plt.title("Scatter Plot")
plt.show()
```

## 4. How do you calculate the correlation matrix using Seaborn and visualize it with a heatmap?

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df=pd.DataFrame({"Math":[70,80,90,60,75],"Science":[65,85,88,55,72],"English":[75,78,92,62,70]})
corr=df.corr()
sns.heatmap(corr, annot=True)
plt.title("Correlation Heatmap")
plt.show()
```

## 5. Generate a bar plot using Plotly.

```python
import matplotlib.pyplot as plt
x=[1,2,3,4,5]
y1=[2,4,6,8,10]
y2=[1,3,5,7,9]
plt.plot(x,y1,label="Line 1")
plt.plot(x,y2,label="Line 2")
plt.legend()
plt.title("Multiple Line Plot")
plt.show()
```
